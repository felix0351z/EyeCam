### Laden von Bild/Videomaterial
Um eine Erkennung von Objekten in einem Bild zu ermöglichen, muss ein Bild / Video zuerst eingelesen werden.
Dabei muss auf das entsprechende Eingabeformat geachtet werden.

### OpenCV
OpenCV ermöglicht das Aufnehmen von Videos mit der Klasse `cv2.VideoCapture`. Mit der Funktion `read()` wird ein Frame (Bild) pro Tick eingelesen. Das `frame` welches als Rückgabe kommt, ist ein 3-Dimensionales Array, welches Höhe, Breite und Farbkanäle beinhaltet (Height, Width, Channel) auch HWC genannt. Channel hat eine Größe von 3 Feldern für die Farben RGB. Hier ist außerdem bei OpenCV die Besonderheit, dass er die Farben als BGR einliest. Dies ist historisch bedingt

https://opencv.org/color-spaces-in-opencv/





```python
import sys
import cv2

# Open the default camera
cam = cv2.VideoCapture(0)

while True:
    success, frame = cam.read()
    if not success:
        sys.exit("Failure: Unable to capture video. Please verify your camera settings.")

    # Display the captured frame
    cv2.imshow('Camera', frame)

    # Format HWC / Height, Width, Channel
    print("Array-Dimensions: ", frame.shape[:2])

    # Press 'q' to exit the loop
    if cv2.waitKey(1) == ord('q'):
        break

# Release the capture and writer objects
cam.release()
cv2.destroyAllWindows()
```

### Picamera2
Um auf dem Pi Videomaterial einzulesen, kann die Library PiCamera2 verwendet werden.
Die Verwendung ist ähnlich zu OpenCV. Da Picam standardmäßig das RGB-Format nutzt, muss falls erwünscht das Format explizit zu BGR geändert werden. Dazu wird das Attribut `format` auf `RGB888` gesetzt.


```python
from picamera2 import Picamera2

picam = Picamera2()
config = picam.create_preview_configuration(
    main={"size": (640, 640), "format": "RGB888"}
)
picam.configure(config)
picam.start()

while True:
    frame = picam.capture_array()

    # Display the captured frame
    cv2.imshow('Camera', frame)

    # Press 'q' to exit the loop
    if cv2.waitKey(1) == ord('q'):
        break

# Release the capture and writer objects
picam.stop()
cv2.destroyAllWindows()
```
