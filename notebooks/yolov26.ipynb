```python
from src.processors.processor import YOLOv8Processor
import cv2 as cv

image = cv.imread('DSC03937.JPG')
p = YOLOv8Processor()
p.initialize()
result = p.process(image)

# YOLOv26 has a built-in NMS (Non-Maximum-Suppression)
# Because of that the output of the model is quite different to older model like the v5
#
# The ndarray has 3 dimensions, 1) the image itself, 2) the detections, 3) the attributes for one detection
# https://docs.ultralytics.com/guides/end2end-detection#using-custom-inference-code
s = result[0].shape
print(f"{s[0]} Frames, {s[1]} Detections, {s[2]} Attributes")

# Now the 6 attributes are [x1, y1, x2, y2, confidence, class_id]
# Lets see how many detections have a confidence over 0.5
detections = result[0][0]
confidences = detections[:, 4]
valid_confidences = confidences >= 0.5

print(f"{valid_confidences.sum()} valid detections")
```
