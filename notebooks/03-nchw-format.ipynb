### Formatierung für die Eingabe in YOLO
Das Bild liegt nun in dem Format HWC vor (Height, Channel, Width). Für die Eingabe in YOLO muss das Bild vorher noch weiterverarbeitet werden.


#### 1. Normalisierung und Umwandlung zu float32
RGB-Farben werden als byte bzw. uint8 angegeben und können einen Wert von 0 - 255 darstellen. Die meisten Deep-Learning-Modelle arbeiten jedoch mit Floats, welche eine Größe von 0 - 1 aufweisen.

#### 2. Umwandlung in NCHW
Zusätzlich erwarten die meisten Modelle das Format NCHW. N ist die erste Achse und steht für die Anzahl der Bilder. Danach muss bereits die Channel-Achse kommen. Das Array muss daher transponiert werden.

#### 3. Konvertierung von BGR zu RGB
Da die Modelle auch mit RGB-Werten arbeiten muss vorher die Channel-Achse wieder geändert werden.



```python

import numpy as np
import cv2 as cv

from notebooks.utils import letterbox

image: np.ndarray = cv.imread('pictures/bicycle.jpg')
resized, transform = letterbox(image)

# Both of these Methods can be done with OpenCV's blobFromImage function.
# This function creates an array with the format NCHW.
# It also supports rescaling, but this is not necessary here since the letterbox function does this.

blob = cv.dnn.blobFromImage(
    image=resized,        # Use the resized image as input
    scalefactor=1 / 255,  # Normalize to
    size=(640, 640),      # Size required by the model
    swapRB=True,          # Convert BGR to RGB
    crop=False            # Cropping
)

print(blob.shape)

```

```python
# Here the same, but manually
import numpy as np
import cv2 as cv


image: np.ndarray = cv.imread('pictures/bicycle.jpg')
resized, transform = letterbox(image)

# 1. Swap BGR to RGB
resized = resized[:, :, ::-1]
# 2. Convert to float
resized = resized.astype("float32")
# 3. Normalizing
resized = resized * (1 / 255)
# 4. HWC -> CHW
chw = resized.transpose(2, 0, 1)
# 5. Add N-Dimension
blob = np.expand_dims(chw, axis=0)

# (1, C, H, W)
print(blob.shape)

```
