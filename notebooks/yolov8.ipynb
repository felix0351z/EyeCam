```python
from src.processors.processor import YOLOProcessor
import cv2 as cv
import numpy as np

image = cv.imread('DSC03937.JPG')
p = YOLOProcessor()
p.initialize("../cnn/yolov8n.onnx")
result = p.process(image)

# YOLOv8 has no built-in NMS (Non-Maximum-Suppression)
# The output structure is different from the v26 model
s = result[0].shape
print(f"{s[0]} Frames, {s[1]} Attributes, {s[2]} Detections")
# Here, the last 2 axes were transposed
# Now the 84 attributes are [center x, center y, width, height, and 80 coco class scores :)]
detections = np.transpose(result[0][0])
class_confidences = detections[:, 4:]
max_confidences = np.max(class_confidences, axis=1)

valid_confidences = max_confidences >= 0.5

print(f"{valid_confidences.sum()} valid detections (No NMS)")
```
